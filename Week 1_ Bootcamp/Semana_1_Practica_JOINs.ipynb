# INNER JOIN
**Ejercicio 2.1: Clientes con sus órdenes**



```sql
SELECT
c.cliente_id,
c.nombre as cliente,
c.ciudad, 
o.orden_id,
o.producto_id,
o.cantidad,
o.fecha

FROM bootcamp.practice.clientes c
INNER JOIN bootcamp.practice.ordenes o ON c.cliente_id = o.cliente_id
ORDER BY c.cliente_id
  
```
